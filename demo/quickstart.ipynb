# NumCompute — Quickstart

This notebook demonstrates the NumCompute toolkit in the required
workflow order: loading, preprocessing, statistics, sorting and
searching, ranking, metrics, gradients, pipelines, and benchmarking.

```python
import numpy as np

from numcompute.io import read_csv
from numcompute.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder
from numcompute.stats import mean, median, std, minimum, maximum, quantile, histogram
from numcompute.sort_search import stable_sort, multi_key_sort, binary_search, top_k, quick_select
from numcompute.rank import rankdata, percentile_rank
from numcompute.metrics import accuracy, confusion_matrix, precision, recall, f1, mse
from numcompute.optim import gradient, jacobian
from numcompute.pipeline import Pipeline

print("NumCompute quickstart")
print("NumPy version:", np.__version__)
print("Imports successful!")
```

## 1. Loading Data

Load a numerical dataset from a CSV file using the toolkit's input/output
module. The loaded array will be used in the following workflow sections.

```python
sample_path = "quickstart_sample.csv"

try:
    data = read_csv(sample_path)
    print("Loaded data:\n", data)
    print("Shape:", data.shape)
except:
    print("Error when reading csv")
```

## 2. Preprocessing

Preprocessing transforms raw numerical and categorical data into
representations suitable for further analysis.

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

print("Original data:\n", data)
print("Standardised data:\n", scaled_data)

minmax_data = MinMaxScaler().fit_transform(data)
print("Min-max scaled data:\n", minmax_data)

categories = np.array([["red"], ["blue"], ["red"]])
encoded = OneHotEncoder().fit_transform(categories)
print("One-hot encoded categories:\n", encoded)
```

## 3. Statistical Analysis

Calculate descriptive statistics to understand the distribution and
central tendency of the loaded dataset.

```python
print("Column means:", mean(data, axis=0))
print("Column medians:", median(data, axis=0))
print("Column standard deviations:", std(data, axis=0))
print("Minimum values:", minimum(data, axis=0))
print("Maximum values:", maximum(data, axis=0))
print("Quartiles:\n", quantile(data, [0.25, 0.5, 0.75], axis=0))

counts, edges = histogram(data[:, 0], bins=4)
print("First-column histogram:", counts)
print("Histogram edges:", edges)
```

## 4. Sorting and Searching

Sort numerical values, search a sorted array, and select the largest
values without manually implementing these algorithms in the notebook.

```python
values = data[:, 0]
sorted_values = stable_sort(values)

print("Values:", values)
print("Sorted values:", sorted_values)
print("Search for 70:", binary_search(sorted_values, 70))

indices = top_k(values, k=2, largest=True)
print("Two largest values:", values[indices])

print(
    "Second-largest value:",
    quick_select(values, k=1, largest=True, return_index=False),
)

order = multi_key_sort(data)
print("Multi-key sorted records:\n", data[order])
```

## 5. Ranking

Ranking assigns relative positions to values. Tied values can receive
average ranks, while descending ranking places the largest value first.

```python
scores = np.array([85, 70, 95, 70, 80])

print("Scores:", scores)
print("Average ranks:", rankdata(scores))
print("Descending ranks:", rankdata(scores, ascending=False))

print(
    "Percentile rank of 80:",
    percentile_rank(scores, 80),
)
```

## 6. Evaluation Metrics

Evaluate example classification predictions and numerical predictions
using the toolkit's metrics functions.

```python
y_true = np.array([1, 0, 1, 1, 0, 1])
y_pred = np.array([1, 0, 1, 0, 0, 1])

print("Accuracy:", accuracy(y_true, y_pred))
print("Confusion matrix:\n", confusion_matrix(y_true, y_pred))
print("Precision:", precision(y_true, y_pred))
print("Recall:", recall(y_true, y_pred))
print("F1 score:", f1(y_true, y_pred))

actual = np.array([2.5, 3.0, 4.5, 5.0])
predicted = np.array([2.0, 3.5, 4.0, 5.5])
print("Mean squared error:", mse(actual, predicted))
```

## 7. Numerical Gradients and Jacobians

Finite differences estimate derivatives numerically. The gradient
describes a scalar-valued function, while the Jacobian describes
the derivatives of a vector-valued function.

```python
def objective(x):
    return x[0] ** 2 + x[1] ** 2

def vector_function(x):
    return np.array([x[0] + x[1], x[0] * x[1]])

point = np.array([3.0, 4.0])

print("Gradient:", gradient(objective, point))
print("Jacobian:\n", jacobian(vector_function, point))
```

## 8. Preprocessing Pipeline

A pipeline chains preprocessing steps so that they can be fitted and
applied in sequence.

```python
pipeline = Pipeline([
    ("standard_scaler", StandardScaler()),
    ("minmax_scaler", MinMaxScaler()),
])

pipeline_data = np.array([
    [10.0, 100.0],
    [20.0, 200.0],
    [30.0, 300.0],
    [40.0, 400.0],
])

pipeline_result = pipeline.fit_transform(pipeline_data)

print("Original data:\n", pipeline_data)
print("Pipeline output:\n", pipeline_result)
print("New data transformed:\n",
      pipeline.transform(np.array([[15.0, 150.0]])))
```
