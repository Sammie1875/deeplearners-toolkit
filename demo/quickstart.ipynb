# DeepLearners Toolkit — Quickstart

This notebook demonstrates the core features of the DeepLearners Toolkit,
a lightweight scientific computing and machine learning utility library
built using Python and NumPy.

We will explore numerical ranking, array validation, and safe numerical operations.

```python
import sys
from pathlib import Path

# Your notebook is inside the demo folder
PROJECT_ROOT = Path.cwd().parent

# Add the project root to Python's module search path
sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
print("Package folder exists:",
      (PROJECT_ROOT / "deeplearners_toolkit").is_dir())
print("Rank file exists:",
      (PROJECT_ROOT / "deeplearners_toolkit" / "rank.py").is_file())
print("Utils file exists:",
      (PROJECT_ROOT / "deeplearners_toolkit" / "utils.py").is_file())
```

```python
import numpy as np

from deeplearners_toolkit.rank import rankdata, percentile_rank
from deeplearners_toolkit.utils import (
    validate_array,
    get_array_info,
    safe_divide,
)

print("DeepLearners Toolkit Quickstart")
print("NumPy version:", np.__version__)
print("Imports successful!")
```

## 1. Ranking Numerical Values

Ranking is used to arrange numerical values according to their relative
positions. The `rankdata()` function supports tied values and ascending
or descending ranking.

In this example, we rank student scores and compare the results.

```python
# 1. Ranking numerical values

scores = np.array([85, 70, 95, 70, 80])

print("Original scores:", scores)
print("Average ranks:", rankdata(scores))
print("Descending ranks:", rankdata(scores, ascending=False))
```

## 2. Percentile Ranking

Percentile ranking helps us understand how a value compares with the
values in a dataset. We use `percentile_rank()` to calculate percentile
ranks for selected scores.

```python
# 2. Percentile ranking

scores = np.array([10, 20, 30, 40, 50])

for score in [20, 30, 40]:
    result = percentile_rank(scores, score)
    print(f"Score {score}: percentile rank = {result:.2f}")
```

## 3. Array Validation and Metadata

Array validation checks whether input data meets the required conditions.
The `get_array_info()` function provides information about an array,
including its shape, dimensions, data type, size, and memory usage.

```python
# Example 3: Array validation and metadata

data = np.array([12.5, 15.0, 18.5, 20.0])

validated_data = validate_array(
    data,
    ndim=1,
    allow_empty=False
)

print("Validated data:", validated_data)
print("Array information:", get_array_info(validated_data))
```

## 4. Safe Numerical Division

The `safe_divide()` function performs numerical division with configurable
handling of zero denominators. In this example, divisions by zero return
NaN instead of raising an exception.

```python
# Example 4: Safe division

numerator = np.array([10, 20, 30])
denominator = np.array([2, 0, 5])

result = safe_divide(
    numerator,
    denominator,
    zero_policy="nan"
)

print("Numerators:", numerator)
print("Denominators:", denominator)
print("Division result:", result)
```

## 5. Error Handling and Input Validation

The toolkit raises clear exceptions when inputs do not meet requirements.
These examples demonstrate handling invalid numerical data and
division-by-zero errors.

```python
# Example 5: Handling invalid inputs

try:
    validate_array([1, 2, np.nan], allow_nan=False)
except ValueError as error:
    print("Validation error:", error)

try:
    safe_divide([10, 20], [2, 0], zero_policy="raise")
except ZeroDivisionError as error:
    print("Division error:", error)
```

## 6. Statistical Analysis

The statistics module provides descriptive statistics for numerical
data, including the mean, median, standard deviation, minimum,
maximum, histogram, and quantiles.

```python

from deeplearners_toolkit.stats import (
    mean,
    median,
    std,
    minimum,
    maximum,
    histogram,
    quantile,
)

data = np.array([10, 20, 30, 40, 50], dtype=float)

print("Dataset:", data)
print("Mean:", mean(data))
print("Median:", median(data))
print("Standard deviation:", std(data))
print("Minimum:", minimum(data))
print("Maximum:", maximum(data))
print("Quartiles:", quantile(data, [0.25, 0.5, 0.75]))

counts, edges = histogram(data, bins=5)
print("Histogram counts:", counts)
print("Histogram edges:", edges)
```

## 7. Sorting and Searching

The `sort_search` module provides algorithms for sorting numerical arrays, sorting rows by multiple keys, searching sorted arrays, and selecting the largest or smallest elements.

This section demonstrates `stable_sort()`, `multi_key_sort()`, `binary_search()`, `top_k()`, and `quick_select()`.

```python
from deeplearners_toolkit.sort_search import (
    stable_sort,
    multi_key_sort,
    binary_search,
    top_k,
    quick_select,
)

# 1. Stable sorting
values = np.array([40, 10, 30, 10, 50, 20])
sorted_values = stable_sort(values)

print("Original values:", values)
print("Sorted values:", sorted_values)

# 2. Sort rows using multiple columns
records = np.array([
    [2, 80],
    [1, 90],
    [2, 70],
    [1, 85],
])

row_order = multi_key_sort(records)

print("\nOriginal records:\n", records)
print("Row sorting indices:", row_order)
print("Sorted records:\n", records[row_order])

# 3. Binary search
search_array = np.array([10, 20, 30, 40, 50])

index, found = binary_search(search_array, 30)
print("\nBinary search for 30:", index, found)

index, found = binary_search(search_array, 35)
print("Binary search for 35:", index, found)

# 4. Select the three largest values
top_indices = top_k(values, k=3, largest=True)
print("\nIndices of three largest values:", top_indices)
print("Three largest values:", values[top_indices])

# 5. Select the second-largest value (zero-based k=1)
second_largest = quick_select(
    values, k=1, largest=True, return_index=False
)
print("Second-largest value:", second_largest)
```

## 8. Data Preprocessing

The `preprocessing` module provides three classes for transforming data: `StandardScaler`, `MinMaxScaler`, and `OneHotEncoder`.

Standard scaling centres numerical features around their means, min-max scaling maps features to a specified range, and one-hot encoding converts categorical columns into numerical indicator columns.

```python
from deeplearners_toolkit.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    OneHotEncoder,
)

# 1. Standard scaling
numerical_data = np.array([
    [10.0, 100.0],
    [20.0, 200.0],
    [30.0, 300.0],
])

standard_scaler = StandardScaler()
standard_scaled = standard_scaler.fit_transform(numerical_data)

print("Original numerical data:\n", numerical_data)
print("\nStandard-scaled data:\n", standard_scaled)

# 2. Min-max scaling
minmax_scaler = MinMaxScaler(feature_range=(0, 1))
minmax_scaled = minmax_scaler.fit_transform(numerical_data)

print("\nMin-max-scaled data:\n", minmax_scaled)

# 3. One-hot encoding
categorical_data = np.array([
    ["Red", "Small"],
    ["Blue", "Large"],
    ["Red", "Large"],
])

encoder = OneHotEncoder()
encoded_data = encoder.fit_transform(categorical_data)

print("\nOriginal categorical data:\n", categorical_data)
print("\nOne-hot encoded data:\n", encoded_data)
print("Encoded shape:", encoded_data.shape)
```

## 9. Model Evaluation Metrics

The `metrics` module provides functions for evaluating classification predictions and numerical predictions.

Classification metrics include accuracy, confusion matrix, precision, recall, and F1 score. Mean squared error (MSE) measures the average squared difference between actual and predicted numerical values.

```python
from deeplearners_toolkit.metrics import (
    accuracy,
    confusion_matrix,
    precision,
    recall,
    f1,
    mse,
)

# Classification example
y_true = np.array([1, 0, 1, 1, 0, 1])
y_pred = np.array([1, 0, 1, 0, 0, 1])

print("Actual labels:", y_true)
print("Predicted labels:", y_pred)

print("\nAccuracy:", accuracy(y_true, y_pred))
print("Confusion matrix:\n", confusion_matrix(y_true, y_pred))
print("Precision:", precision(y_true, y_pred))
print("Recall:", recall(y_true, y_pred))
print("F1 score:", f1(y_true, y_pred))

# Regression example
actual_values = np.array([2.5, 3.0, 4.5, 5.0])
predicted_values = np.array([2.0, 3.5, 4.0, 5.5])

print("\nActual numerical values:", actual_values)
print("Predicted numerical values:", predicted_values)
print("Mean squared error:", mse(actual_values, predicted_values))
```

## 10. Numerical Gradients and Jacobians

The `optim` module estimates derivatives using finite differences. A gradient estimates the derivatives of a scalar-valued function, while a Jacobian represents the derivatives of a vector-valued function.

Both central and forward difference methods are supported by the main `gradient()` and `jacobian()` functions. The module also provides batched variants for functions that support batched inputs.

```python
from deeplearners_toolkit.optim import (
    gradient,
    jacobian,
    gradient_batched,
    jacobian_batched,
)

# Scalar-valued function: f(x) = x[0]**2 + x[1]**2
def objective(x):
    return x[0]**2 + x[1]**2

point = np.array([3.0, 4.0])

estimated_gradient = gradient(objective, point)

print("Input point:", point)
print("Estimated gradient:", estimated_gradient)
print("Expected gradient approximately:", 2 * point)

# Vector-valued function
def vector_function(x):
    return np.array([
        x[0] + x[1],
        x[0] * x[1],
    ])

jacobian_matrix = jacobian(vector_function, point)

print("\nEstimated Jacobian:\n", jacobian_matrix)

# Batched variants are included in the module.
# These examples use functions that accept batches of inputs.
def batched_objective(X):
    return np.sum(X**2, axis=tuple(range(1, X.ndim)))

batched_grad = gradient_batched(batched_objective, point)
print("\nBatched gradient estimate:", batched_grad)

def batched_vector_function(X):
    return np.stack(
        (X[..., 0] + X[..., 1], X[..., 0] * X[..., 1]),
        axis=-1,
    )

batched_jac = jacobian_batched(batched_vector_function, point)
print("\nBatched Jacobian estimate:\n", batched_jac)
```

## 11. Building a Preprocessing Pipeline

The `pipeline` module chains multiple components into a single workflow. Each preprocessing step transforms the output of the previous step, allowing data transformations to be applied in a consistent order.

This example combines standard scaling and min-max scaling in one pipeline using `Pipeline.fit_transform()`.

```python
from deeplearners_toolkit.pipeline import Pipeline

pipeline_data = np.array([
    [10.0, 100.0],
    [20.0, 200.0],
    [30.0, 300.0],
    [40.0, 400.0],
])

preprocessing_pipeline = Pipeline([
    ("standard_scaler", StandardScaler()),
    ("minmax_scaler", MinMaxScaler(feature_range=(0, 1))),
])

transformed_data = preprocessing_pipeline.fit_transform(pipeline_data)

print("Original data:\n", pipeline_data)
print("\nTransformed data:\n", transformed_data)

# Transform new data using the already-fitted pipeline
new_data = np.array([
    [15.0, 150.0],
    [35.0, 350.0],
])

new_transformed_data = preprocessing_pipeline.transform(new_data)

print("\nNew data:\n", new_data)
print("\nTransformed new data:\n", new_transformed_data)
```

## 12. Loading Data from CSV

The `io` module provides `read_csv()`, which loads CSV data into a NumPy array. It supports a configurable delimiter, a replacement value for missing entries, and an output data type.

The following example creates a small sample CSV file, loads it using the toolkit, and removes the temporary file afterwards.

```python
from pathlib import Path
from deeplearners_toolkit.io import read_csv

sample_csv = Path("quickstart_sample.csv")

try:
    # Create a small CSV dataset for demonstration
    np.savetxt(
        sample_csv,
        np.array([
            [10.0, 100.0],
            [20.0, 200.0],
            [30.0, 300.0],
        ]),
        delimiter=",",
        fmt="%.1f",
    )

    loaded_data = read_csv(sample_csv)

    print("Loaded CSV data:\n", loaded_data)
    print("Loaded data shape:", loaded_data.shape)

finally:
    # Avoid leaving a sample file in the project
    if sample_csv.exists():
        sample_csv.unlink()
```

## 13. End-to-End Numerical Workflow

This section combines several toolkit modules in a single example. It loads data from a CSV file, validates the resulting array, calculates descriptive statistics, ranks observations, scales the features, and evaluates sample predictions.

This demonstrates how individual functions can be used together in a practical numerical workflow.

```python
from deeplearners_toolkit.io import read_csv
from deeplearners_toolkit.stats import mean, median, std
from deeplearners_toolkit.rank import rankdata
from deeplearners_toolkit.utils import validate_array, get_array_info
from deeplearners_toolkit.preprocessing import StandardScaler
from deeplearners_toolkit.metrics import mse

# Create a temporary dataset
workflow_csv = Path("quickstart_workflow.csv")

try:
    dataset = np.array([
        [50.0, 60.0],
        [70.0, 80.0],
        [90.0, 100.0],
        [60.0, 75.0],
    ])

    np.savetxt(workflow_csv, dataset, delimiter=",", fmt="%.1f")

    # Step 1: Load and validate
    loaded = read_csv(workflow_csv)
    validated = validate_array(
        loaded,
        ndim=2,
        allow_empty=False,
        allow_nan=False,
    )

    print("Step 1 — Loaded data:\n", validated)
    print("Array metadata:", get_array_info(validated))

    # Step 2: Calculate statistics on the first column
    first_feature = validated[:, 0]

    print("\nStep 2 — First-column mean:", mean(first_feature))
    print("First-column median:", median(first_feature))
    print("First-column standard deviation:", std(first_feature))

    # Step 3: Rank the first column
    print("\nStep 3 — First-column ranks:", rankdata(first_feature))

    # Step 4: Scale the features
    scaler = StandardScaler()
    scaled = scaler.fit_transform(validated)

    print("\nStep 4 — Standard-scaled data:\n", scaled)

    # Step 5: Evaluate example numerical predictions
    actual = np.array([10.0, 20.0, 30.0])
    predicted = np.array([12.0, 18.0, 33.0])

    print("\nStep 5 — Example prediction MSE:", mse(actual, predicted))

finally:
    if workflow_csv.exists():
        workflow_csv.unlink()
```
